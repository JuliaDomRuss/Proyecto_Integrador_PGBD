```python
import os
import yfinance as yf

# period="max" descarga desde el primer día que hay registro
sp500 = yf.download(
    "^GSPC",
    start="1990-01-01",
    auto_adjust=False,
    multi_level_index=False,
)

print(sp500.head())
print(sp500.tail())
print(sp500.shape)

os.makedirs("../data/raw", exist_ok=True)
sp500.to_csv("../data/raw/s&p500_diario.csv", index=False)
```
