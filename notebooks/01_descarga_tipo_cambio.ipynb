## Descarga de BD de Banxico sobre tipo de cambio a través de API

```python
import os
from datetime import date
import requests
import pandas as pd

TOKEN = os.environ["BANXICO_TOKEN"]          # lo lee de la computadora, no del código
SERIE = "SF43718"                            # Tipo de cambio FIX
INICIO = "1991-11-12"
FIN = date.today().isoformat()

url = f"https://www.banxico.org.mx/SieAPIRest/service/v1/series/{SERIE}/datos/{INICIO}/{FIN}"
resp = requests.get(url, headers={"Bmx-Token": TOKEN, "Accept": "application/json"}, timeout=60)
resp.raise_for_status()                      # si algo falla, aquí te avisa

datos = resp.json()["bmx"]["series"][0]["datos"]
df = pd.DataFrame(datos)                     # columnas: fecha (dd/mm/aaaa), dato (texto)

print("Registros:", len(df))
print("Sin dato (N/E):", (df["dato"] == "N/E").sum())
print(df.head(3), df.tail(3), sep="\n")

os.makedirs("../data/raw", exist_ok=True)
df.to_csv("../data/raw/tipo_cambio_fix.csv", index=False)
```
